```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from IPython.display import display, clear_output
```

```python
# Generate synthetic data
"""
1. np.random.seed(0)

Sets the seed for NumPy’s random number generator.

Ensures that any random numbers generated after this line are reproducible.

Example: if you call np.random.rand() multiple times after this seed, you will always get the same sequence of “random” numbers.

2. X = np.linspace(-5, 5, 100)

Generates 100 evenly spaced numbers between -5 and 5.

Syntax: np.linspace(start, stop, num_points)

start = -5 → first number

stop = 5 → last number

num_points = 100 → total numbers generated
"""
np.random.seed(0)
X = np.linspace (-5, 5, 100)
```

```python
# Create a function for interactive visualization
"""
Converts probabilities into binary labels:

If new_probability > threshold → label = 1

Else → label = 0

.astype(int) converts True/False to 1/0.

 Linear Combination: This plot shows the relationship bewteen the feature values X and their linear
 combination.

 Logistic function(probabilities): this plot shows the sigmoid curve, wich maps the linear combination
 values to probabilities between 0 and 1

 Predicted labels: This plot shows the final binary classification (0 or 1) for each data point, based
 on the threshold applied to the probabilities.
"""
def plot_logistic_regression(slope=2.0, intercept = -0.1, threshold=0.5):
    new_linear_combination = slope * X + intercept
    new_probability = 1 / (1 + np.exp(-new_linear_combination))
    predicted_labels = (new_probability > threshold).astype(int)

    # Create subplots
    fig, (ax1, ax2, ax3) = plt.subplots(1,3,figsize=(18,5))

    # Plot the linear combination
    ax1.plot(X, new_linear_combination, label='Linear Combination')
    ax1.set_xlabel('Feature Value (X)')
    ax1.set_ylabel('Linear Combination (y)')
    ax1.legend()

    # Plot the logistic function output as a subplot
    ax2.plot(X, new_probability, label='Logistic Function')
    ax2.set_xlabel('Feature Value (X)')
    ax2.set_ylabel('Probability')
    ax2.legend()

    #Plot the predicted labels
    ax3.scatter(X, predicted_labels, label='Predicted Lables')
    ax3.set_xlabel('Feature Value(X)')
    ax3.set_ylabel('Predicted Label')
    ax3.legend()

    plt.show()
    
```

```python
# Create sliders
slope_slider = FloatSlider(value=2.0, min=-0.5, max=5.0, step=0.1, description='Slope')

intercept_slider = FloatSlider(value=-0.1, min=-0.5, max=5.0, step=0.1, description='Intercept')

threshold_slider = FloatSlider(value=0.5, min=0.5, max=1.0, step=0.01, description='Threshold')

# Create an interactive widget
interact(plot_logistic_regression, slope=slope_slider, intercept=intercept_slider, threashold=threshold_slider)
```
