# Clustering
- Unsupervised learning techinque that groups similar data points into clusters based on their **feature**.
- The **goal** is to identify patterns or structures in the data without prior knowledge of class labels.
- Ex:
- **Customer Segmentation**: Group customers by demographics, behavior, or prefernces to target marketing efforts.
- **Image Segmentation**: Cluster pixels in an image to identify objets or regions.
- **Gene Expression Analysis**: Cluster genes based on their expression levels to identify co-regulated genes.
- **Anomaly Detection**: Identify outliers or unusual patterns in data using clustering algorithms.

### KMeans Algorithm
**KMeans** is a popular clustering algorithm that partitions the data into K clusters based on their mean distances. The algorithm iteratively updates the cluster centroids (mean points) and reassigns data points to the closest cluster until convergence.

### How KMeans Works
    1. Initialize K cluster centroids randomly.
    2. Assign each data points to the closest centroid based on Euclidean distance.
    3. Update the centroid of each cluster by calculating the mean of all data points assigned to it.
    4. Repeat steps 2-3 until the centroids converge or a stopping criterion is reached.

- **Increasing** the number of clusters (K)** allows for more detailed clustering, but may lead to overfitting or noisy clusters.
- **Decreasing** K may result in underfitting or loss of important patterns,
- **The centroids** represent the mean points of each cluster. As the number of clsuters changes, the centroids adjust to reflect the new cluster assignments. 



```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn import datasets
from ipywidgets import interact, IntSlider
```

```python
# Load the Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2] # Use only the first two features for simplicity
```

```python
def plot_kmeans_clusters(num_clusters=2):
    # 1: create the KMeans object
    kmeans = KMeans(n_clusters=num_clusters, random_state=0)
    # 2: fit and get labels
    labels = kmeans.fit_predict(X)

    # 3: prepare the figure
    plt.figure(figsize=(10,6))

    # 4: plot each cluster's points (one call per cluster)
    for i in range(num_clusters):
        plt.scatter(
            X[labels == i, 0],            
            X[labels == i, 1],            
            s=50,                         
            alpha=0.6,                    
            label=f'Cluster {i + 1}'      
        )

    # 5: plot centroids ONCE (outside the loop)
    plt.scatter(
        kmeans.cluster_centers_[:, 0],   # centroid x coordinates
        kmeans.cluster_centers_[:, 1],   # centroid y coordinates
        s=200,                           # big marker to stand out
        c='black',                       # color
        marker='X',                      # shape
        label='Centroids'                # legend label
    )

    # 6: labels, title and legend (one time)
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title(f'K-Means Clustering (Number of Clusters: {num_clusters})')
    plt.legend()
    plt.grid(True)
    # 7: display the finished figure
    plt.show()
        
```

```python
# Create a slider for adjusting the number of clusters
num_clusters_slider = IntSlider(value=2, min=2, max=5, description='Number of Clusters')

# Create an interactive widget
interact(plot_kmeans_clusters, num_clusters=num_clusters_slider)
```
